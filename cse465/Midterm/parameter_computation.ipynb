```python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),                                   # grayscale image, e.g. MNIST
    layers.Conv2D(8, kernel_size=3, activation="relu", name="conv1"),
    layers.MaxPooling2D(pool_size=2, name="pool1"),
    layers.Conv2D(16, kernel_size=3, activation="relu", name="conv2"),
    layers.MaxPooling2D(pool_size=2, name="pool2"),
    layers.Flatten(name="flatten"),
    layers.Dense(32, activation="relu", name="dense1"),
    layers.Dense(10, activation="softmax", name="output"),
])
model.summary()
```

```python
model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(8, kernel_size=3, strides=2, padding="valid", activation="relu", name="conv1"),
    layers.MaxPooling2D(pool_size=3, strides=2, padding="same", name="pool1"),
    layers.Conv2D(16, kernel_size=3, strides=1, padding="same", activation="relu", name="conv2"),
    layers.MaxPooling2D(pool_size=2, strides=2, padding="valid", name="pool2"),
    layers.Flatten(name="flatten"),
    layers.Dense(32, activation="relu", name="dense1"),
    layers.Dense(10, activation="softmax", name="output"),
])
model.summary()
```
